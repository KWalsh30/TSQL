```python
import random

def check_guess(guess, answer):
    if guess < answer:
        return "📉 Too low!"
    elif guess > answer:
        return "📈 Too high!"
    else:
        return "🎉 Correct!"

def play_game():
    answer = random.randint(1, 100)
    attempts = 0
    print("\n🎮 Guess a number between 1 - 100!")

    while True:
        try:
            guess = int(input("Your guess: "))
            attempts += 1
            print(check_guess(guess, answer))
            if guess == answer:
                print(f"Got it in {attempts} attempt(s)!")
                break
        except ValueError:
            print("⚠️ Enter a valid number!")

def main():
    while True:
        play_game()
        if input("\nPlay again? (yes/no): ").lower() != "yes":
            print("Thanks for playing! 👋")
            break

main()
```
